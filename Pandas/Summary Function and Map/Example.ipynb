## Maps <br>
In simple terms, the maps help us to reformat, transform, and modify the data into the format we need by applying function. Instead of changing each value maually.
For example, we might use the maps in these situations:
 - Replace values with meaningful labels like A -> Apple or vice versa
 - Unit conversions like currency, temperature, weights, etc.
 - Easily apply a function to an entire dataset.



There are two mapping methods we will use often:
1. `map()`
2. `apply()`

For example, suppose we wanted to remean the scores the wines received to 0 &nbsp; (To more easily observe each value differ to the mean) : 

```python
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
```

First, we store the mean of the wines' points into the variables: &nbsp; `review_points_mean`

The second line transforms each value in the `reviews.points` column by subtracting the mean score. 

This method is called ***mean centering or remeaning***

`.map(lambda p: p - review_points_mean)` -> This applies a function to each value `p` in the `points` colums:
 - `p` represents each individual wine score
 - `lambda` a quick, one-time-use function inside the `map()`

```python
reviews
```

```python
result = reviews['points'] - review_points_mean
result
```

```python
def remean_points(row):
    row.points = row.points - review_points_mean
    return(row)

reviews.apply(remean_points, axis = 'columns')
```
