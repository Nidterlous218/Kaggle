## Indexing

Those methods above for indexing in data frame are nice since the principles are same as the Python and it is easy to pick up and use for novice like myself. 

However, Pandas has its own operators for indexing, which is &nbsp; `loc` &nbsp; and &nbsp; `iloc`. These are the advanced operators we are supposed to using.

```python
# Indexing
```

### `iloc[ ]`

The &nbsp; `iloc` &nbsp; selects data based on the data's numerical position, also known as the ***index - based selection***.

**Appendix:** The name of &nbsp; `iloc` &nbsp; is : &nbsp; ***Integer - location based indexing.***

```python
reviews.iloc[0]
```

Typing `reviews.iloc[0]` will select the first row of the data frame.

But why isn't the output should be something like the full value of the column 0 or something else?

Because `iloc` is &nbsp; ***row - first, column - second***. Which is the opposite of the Python :  &nbsp; ***column - first, row - second***.

This means that &nbsp; `iloc`'s &nbsp; **first priority** is to retireve data from data frame's rows, then columns.

Let's take a look below:

```python
reviews.iloc[:, 0]
```

The colon `:` inside the square bracket `[ ]` means ***select all rows***.

In a nutshell, the code &nbsp; `reviews.iloc[:, 0]` &nbsp; means select all rows value of the first column ( 0 )

```python
reviews.iloc[:3, 0]
```

In this code, the `:3` means select rows until the `set_number - 1`.

In this example, the code will retrieve the &nbsp;  ***first (0), second (1), and third (2)*** &nbsp; rows of datas from column 0

What happens if it is &nbsp; `number_a : number_b` &nbsp; ?

It will retrieve the rows start from ***number_a until number_b - 1***

As the example below:

```python
reviews.iloc[1:3, 0]
```

" 1, 2, .....3 and stop, no more data starting from 3"

We could also pass list inside:

```python
reviews.iloc[[0, 1, 2], 0]
```

```python
reviews.iloc[0, [0, 1, 2]]
```

**Note:** To find the last n elements of the data, we could use &nbsp; `-` &nbsp; to represent.

-5 means the last 5 elements

```python
reviews.iloc[-5:]
```

### `loc[ ]`

The second method for attribute selection is the &nbsp; `loc` &nbsp; operator: &nbsp; ***Label - based selection***

It is the data index value, rather than the position

```python
reviews.loc[0, 'country']
```

`iloc` is much simpler than `loc` because it ignores the dataset's indices. &nbsp; We use `iloc` to treat a dataset like a big lists, the only thing we should consider is the position.
On the other hand, `loc` uses the information in the indices. It is usually easier than `iloc` when you have specific indices.

For example, here's one situation where using `loc` is much more easier:

```python
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```

If we need to use `iloc` to produce the same results, it would be like: 

```python
reviews.iloc[:, [8, 9, 3]]
```

This takes more ***work*** to learn where exactly is those columns in positions.

**Note:** When using `loc` and `iloc`, there's one thing to keep in mind which made these two methods slightly different.
          

`iloc[0:10]` will select entries `0,....,9`, while `loc[0:10]` will select entries `0,....,10`

The core idea are `loc` can include strings. Therefore, we hope to index the required range of entries conveniently. 

If we want to select a certain ranges of alphabet between ***a ~ m***. &nbsp; It is more convenient to use `df.loc[a:m]` rather than `df.loc[a:n]`. (**n** come after **m**)

It is fun and confusing when the data frame is a simple numerical list:

For example, a data frame contains `0,......,1000`.

`df.iloc[0:1000]` will return 1,000 entries, while `df.loc[0:1000]` will return 1,001 entries.    >_<

To get exactly 1,000 entries using `loc`, you will need to lower one digit down and change into: `df.loc[0:999]`

***Because unlike humans, computers always start counting from 0.***

## Selection

Selection with conditional is like a ***filter*** to help us select the needed data sets from data frame.

For example, suppose we are interested specifically in better - than average wines produced in Italy.

We could start by checking if each wine is made in italy or not: 

```python
reviews.country == 'Italy'
```

This operation would return a Series of `True / False` based on the condition matched to the `country` records.

This result could be used inside of the `loc` to select the relevant data:

```python
reviews.loc[reviews.country == 'Italy']
```

The code tells the interpreter: "Hey, please help me to find the data that are relevant to the country Italy."

We also wanted to know the better wines with at least 90 points rated:

```python
reviews.loc[(reviews.country == 'Italy') & (reviews.points >= 90)]
```

By using the `&` will bring the two conditions together. &nbsp; The `&` means `and` in the boolean

Suppose we'll buy any wine that's made in Italy `or` rated above 90 points.

For this, we use a pipe (`|`). &nbsp; This means `or` in the boolean    

```python
reviews.loc[(reviews.country == 'Italy') | (reviews.points >= 90)]
```

Pandas comes with a few built-in conditional selectors, two of which will highlight here: &nbsp; `isin` &nbsp; & &nbsp; `notnull`

`isin` lets you select data whose values "is in" a list of values. 

For example, we can use it to select wines only from ***Italy or France***:

```python
reviews.loc[reviews.country.isin(['Italy', 'France'])]
```

Using the boolean condition would achieve the same results, but more complex. 

```python
reviews.loc[(reviews.country == 'Italy') | (reviews.country == 'France')]
```

That's why using the `isin` would be more easier. Imagine you have more than two countries to list out......

The second is `notnull`. It lets you highlight the values which aren't &nbsp; ***empty or not available &nbsp; (NaN values)***. Vise versa for `isnull`

For example, to filter out wines lacking a price tag in the dataset, here's how we could do:

```python
reviews.loc[reviews.price.notnull()]
```

"Please list out those wines that have price tags."
