`groupby()` helps you ***group similar rows based on the values in a specific column (or columns)*** and apply a function like (`sum()`, `mean()`, `count()`, etc.) to summarize the data.

There are four types of examples about the use cases of `groupby()`:
1. Group by a Single Column
2. Group by Multiple Columns
3. Applying different aggregations : &nbsp; `.agg()`
4. Reset Index for Better Readibility : &nbsp; `.reset_index()`



```python
# Grouping
reviews.groupby('points').points.count() # Groupby the column 'points', and calculate how many times it appeared
```

This said: *"Hey, please help me to create a Series by grouping the rows based on the `points` column and counting how many times each unique `points` value appears."*

The code upon is similar to the `value_counts()` as we learn before from the ***Summary function*** section.

By default, the `groupby()` function will sort the index in ***ascending order***, to disable it just simply add `sort = False` behind.

```python
reviews.points.value_counts()
```

If we would like to group the data with the minimum prices:

```python
reviews.groupby('points').price.min()
```

Of course, we could group by ***multiple columns***, the more we add, the more details:

```python
reviews.groupby(['points', 'country']).price.min()
```

```python
reviews.groupby(['points', 'country', 'variety']).price.max()
```

Using the `apply()` method, we could manipulate data in any way neeed.

For example, selecting the name of the first wine reviewed from each winery in the dataset :

```python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0]) # groupby the category 'winery', then in each type of winery, shows the first row of it
```

Since in the dataset contains same winery but with various title. At here, we just select the first title among the same winey category.

Let's look into another example with more complicated.

For an example, we would like to learn details about the highest point of a winery within a province inside a country:

```python
reviews.groupby(['country', 'province']).apply(lambda x: x.loc[x.points.idxmax()])
```

Besides the function `.apply()` , there's another function works well with the `groupby()` function: &nbsp; `.agg()`

The aggregation function (`.agg()`) summarize the data (The previous section about the summary):
- `sum()`
- `mean()`
- `max()`
- `min()`
- `count()`
- `std()`
- .....

Let's say we would like to summarize out the len, min, and max of the price's data and groupby country:

```python
reviews.groupby('country').price.agg([len, min, max])
```

```python
country_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
country_reviewed
```

```python
mi = country_reviewed.index
type(mi)
```

After using the `groupby()` function, the grouped column becomes the indes (in this example: ***country & province***). &nbsp; To restore it as a normal column, use `reset_index()`:

```python
country_reviewed.reset_index()
```

In a nutshell, the `groupby()` function provides us a more detailed description about the data associated with the aggregation function.

## Sorting

In the groupby() function, we have mentioned that the default sorting is ascending in the index order. From this section,  we would learn to sort the data frame with our desired.
1. `sort_values()`
2. `sort_index()`

```python
country_reviewed 
```

```python
country_reviewed = country_reviewed.reset_index()
country_reviewed.sort_values('len')
```

To sort from largest to smallest, set `ascending = False`

```python
country_reviewed.sort_values('len', ascending = False)
```

If we want to sort rows based on the index instead of the values, use `sort_index()`

```python
country_reviewed.sort_index()
```

If we would like to sort simultaneously with two columns:

```python
country_reviewed.sort_values(['country', 'len'])
```
